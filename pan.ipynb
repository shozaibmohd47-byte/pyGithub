### filtering data 

```python
data = {
"id": [1,2,3,4,5],
"name":["shazab","ammar","zayed","awais","khan"],
"marks":[38,73,55,33,52],
'age':[22,32,44,33,19],
"department":["IT","HR","HR","IT","IT"]
}
df = pd.DataFrame(data,index=['a','b','c','d','e'])
df
```

```python
df[df['marks']>50]
```

```python
df.loc[df['marks']>50,['name','age']]
```

```python
import pandas as pd
```

```python
df.loc[df['marks']>50,['name','age']]
```

```python
df.loc[df['name'].str.startswith('a'),['id','name']]
```

```python
df [(df['marks']>50)&(df['department']=='IT')]
```

```python
df [(df['marks']>50)&(df['department'].isin(['IT','HR']))]
```

```python
df [(df['marks']>50)&~(df['department'].isin(['IT']))]
```

```python
df[df['marks'].between(20,50)]
```

```python
df.loc[df['marks'].between(20,50),['id','name']]
```
